```python
# import models
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
```

```python
model_name = "dineshpiyasamara/Llama-2-7b-hf-sentiment-analysis-new"

# load the tokenizer
tokenizer = AutoTokenizer.from_pretrained(model_name)

# load the tokenizer
model = AutoModelForCausalLM.from_pretrained(
    model_name,
    device_map="auto"
)
```

```python
# initialize a text generation pipeline using the fine-tuned model and tokenizer
pipe = pipeline(task="text-generation", model=model, tokenizer=tokenizer, max_length=200)
```

```python
prompt = "I like that product"

prompt_template=f'''[INST] <<SYS>>
Analyze the sentiment of the following text and return either 'positive' or 'negative'
<</SYS>>
{prompt}[/INST]
'''
result = pipe(prompt_template)
print(result[0]['generated_text'])
```
